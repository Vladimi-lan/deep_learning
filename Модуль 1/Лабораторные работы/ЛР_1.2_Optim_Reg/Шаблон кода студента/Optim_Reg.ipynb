```python
# -*- coding: utf-8 -*-
"""
Шаблон для выполнения ЛР 1.2 "Эксперименты с оптимизаторами и регуляризацией"

Инструкция:
1. Заполните все места, помеченные # TODO, реализовав соответствующие алгоритмы.
2. Запустите код для проведения экспериментов и построения графиков.
3. Проанализируйте результаты и подготовьте отчёт.

Все реализации должны быть выполнены только с использованием NumPy.
"""

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical

# ======================================================================
# 1. ЗАГРУЗКА И ПОДГОТОВКА ДАННЫХ (не требует изменений)
# ======================================================================

def load_and_prepare_data():
    """
    Загружает CIFAR-10, нормализует и преобразует в плоские векторы.
    Возвращает обучающую и тестовую выборки.
    """
    (X_train, y_train), (X_test, y_test) = cifar10.load_data()
    X_train = X_train.reshape(X_train.shape[0], -1) / 255.0
    X_test  = X_test.reshape(X_test.shape[0], -1) / 255.0
    y_train = y_train.flatten()
    y_test  = y_test.flatten()

    # Стандартизация (улучшает сходимость)
    mean = X_train.mean(axis=0)
    std = X_train.std(axis=0) + 1e-8
    X_train = (X_train - mean) / std
    X_test = (X_test - mean) / std

    return X_train, y_train, X_test, y_test


# ======================================================================
# 2. ОПТИМИЗАТОРЫ (реализовать самостоятельно)
# ======================================================================

class SGD:
    """
    Стохастический градиентный спуск.
    Формула обновления: w = w - lr * grad
    """
    def __init__(self, learning_rate=0.01):
        # TODO: сохранить learning_rate
        pass

    def update(self, params, grads):
        """
        Обновляет параметры модели.
        params, grads — словари с ключами 'W0', 'b0', 'W1', 'b1', ...
        """
        # TODO: для каждого ключа выполнить обновление w -= lr * grad
        pass


class SGDMomentum:
    """
    SGD с моментом.
    Формулы: v = mu * v - lr * grad; w = w + v
    """
    def __init__(self, learning_rate=0.01, momentum=0.9):
        # TODO: сохранить параметры и инициализировать словарь для хранения скоростей (velocities)
        pass

    def update(self, params, grads):
        # TODO: для каждого ключа обновить скорость и параметры по формулам
        pass


class RMSProp:
    """
    RMSProp оптимизатор.
    Формулы: cache = beta * cache + (1-beta) * grad^2; w = w - lr * grad / (sqrt(cache) + eps)
    """
    def __init__(self, learning_rate=0.001, beta=0.9, eps=1e-8):
        # TODO: сохранить параметры и инициализировать словарь для кэша
        pass

    def update(self, params, grads):
        # TODO: для каждого ключа обновить кэш и параметры
        pass


class Adam:
    """
    Adam оптимизатор (с коррекцией смещения).
    Формулы:
        m = beta1 * m + (1-beta1) * grad
        v = beta2 * v + (1-beta2) * grad^2
        m_hat = m / (1 - beta1^t)
        v_hat = v / (1 - beta2^t)
        w = w - lr * m_hat / (sqrt(v_hat) + eps)
    """
    def __init__(self, learning_rate=0.001, beta1=0.9, beta2=0.999, eps=1e-8):
        # TODO: сохранить параметры, инициализировать словари m, v и счётчик шагов t
        pass

    def update(self, params, grads):
        # TODO: обновить m, v, выполнить коррекцию смещения и обновить параметры
        pass


# ======================================================================
# 3. СЛОЙ С РЕГУЛЯРИЗАЦИЕЙ (реализовать самостоятельно)
# ======================================================================

class Layer:
    """
    Полносвязный слой с поддержкой:
    - активаций: relu, softmax, linear
    - регуляризации L1/L2 (добавляется к градиенту)
    - Dropout (во время обучения случайно обнуляет нейроны)
    """
    def __init__(self, n_input, n_output, activation='relu', dropout_prob=0.0,
                 l2_lambda=0.0, l1_lambda=0.0):
        # TODO: инициализировать веса (He для ReLU, иначе малые случайные) и смещения
        # Сохранить параметры регуляризации и активации
        # Инициализировать поля для хранения промежуточных значений (input, Z, output, dropout_mask)
        pass

    def forward(self, X, is_training=True):
        """
        Прямой проход.
        :param X: вход (batch_size, n_input)
        :param is_training: флаг обучения (для dropout)
        :return: выход слоя
        """
        # TODO: вычислить Z = X·W + b, сохранить в self.Z
        # TODO: применить активацию (relu, softmax, linear)
        # TODO: если is_training и dropout_prob > 0, применить dropout (создать маску и масштабировать)
        # Сохранить вход и выход для обратного прохода
        pass

    def backward(self, dout):
        """
        Обратный проход: вычисляет градиенты и возвращает их, не обновляя параметры.
        :param dout: градиент по выходу (dLoss/dOutput)
        :return: (dX, dW, db) — градиенты по входу, весам и смещениям
        """
        # TODO: вычислить dZ (производную активации)
        # TODO: применить маску dropout (если была)
        # TODO: вычислить dW и db (усреднить по батчу)
        # TODO: добавить регуляризацию L1 и L2 к dW (если заданы коэффициенты)
        # TODO: вычислить dX = dZ·W^T
        # Вернуть dX, dW, db
        pass


# ======================================================================
# 4. МНОГОСЛОЙНЫЙ ПЕРЦЕПТРОН (MLP)
# ======================================================================

class MLP:
    """
    Многослойный перцептрон с поддержкой различных оптимизаторов и регуляризации.
    """
    def __init__(self, layer_sizes, activations, dropout_probs=None,
                 l2_lambda=0.0, l1_lambda=0.0, optimizer='sgd', **opt_kwargs):
        """
        :param layer_sizes: список размеров слоёв, например [3072, 256, 128, 10]
        :param activations: список активаций для каждого слоя
        :param dropout_probs: список вероятностей dropout для каждого слоя (кроме выходного)
        :param l2_lambda: глобальный коэффициент L2
        :param l1_lambda: глобальный коэффициент L1
        :param optimizer: тип оптимизатора ('sgd', 'sgd_momentum', 'rmsprop', 'adam')
        :param opt_kwargs: параметры для оптимизатора (learning_rate, momentum, beta и т.д.)
        """
        # TODO: создать список слоёв (экземпляров Layer) с переданными параметрами
        # TODO: создать оптимизатор (по имени optimizer) с переданными параметрами
        # TODO: собрать словари params и grads для использования в оптимизаторе
        pass

    def forward(self, X, is_training=True):
        # TODO: последовательно применить все слои, вернуть выход последнего
        pass

    def backward_and_update(self, dout, learning_rate):
        """
        Выполняет обратный проход, собирает градиенты, затем обновляет параметры через оптимизатор.
        """
        # TODO: выполнить обратный проход через все слои (в обратном порядке), собрать градиенты в словарь grads
        # TODO: обновить параметры через оптимизатор (self.optimizer.update(self.params, grads))
        # TODO: обновить веса и смещения в слоях из self.params
        pass

    def train_step(self, X_batch, y_batch, learning_rate):
        """
        Один шаг обучения: forward, loss, backward, update.
        """
        # TODO: прямой проход (is_training=True) -> probs
        # TODO: вычислить кросс-энтропию (среднюю по батчу)
        # TODO: вычислить градиент для выходного слоя: dout = (probs - y_one_hot) / batch_size
        # TODO: вызвать backward_and_update
        # Вернуть значение потерь
        pass

    def predict(self, X):
        # TODO: прямой проход (is_training=False), вернуть argmax
        pass

    def accuracy(self, X, y):
        # TODO: сравнить предсказания с истинными метками, вернуть долю правильных
        pass


# ======================================================================
# 5. ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ (не требуют изменений)
# ======================================================================

def train_model(model, X_train, y_train, X_val, y_val,
                batch_size=64, epochs=30, learning_rate=0.001, verbose=True):
    """
    Обучает модель и возвращает историю потерь и точности.
    """
    num_train = X_train.shape[0]
    num_batches = int(np.ceil(num_train / batch_size))
    loss_history = []
    train_acc_history = []
    val_acc_history = []

    for epoch in range(epochs):
        indices = np.random.permutation(num_train)
        X_shuffled = X_train[indices]
        y_shuffled = y_train[indices]

        epoch_loss = 0.0
        for batch_idx in range(num_batches):
            start = batch_idx * batch_size
            end = min(start + batch_size, num_train)
            X_batch = X_shuffled[start:end]
            y_batch = y_shuffled[start:end]

            loss = model.train_step(X_batch, y_batch, learning_rate)
            epoch_loss += loss * (end - start)

        epoch_loss /= num_train
        loss_history.append(epoch_loss)

        train_acc = model.accuracy(X_train, y_train)
        train_acc_history.append(train_acc)
        val_acc = model.accuracy(X_val, y_val)
        val_acc_history.append(val_acc)

        if verbose:
            print(f"Epoch {epoch+1}/{epochs}  Loss: {epoch_loss:.4f}  Train Acc: {train_acc:.4f}  Val Acc: {val_acc:.4f}")

    return loss_history, train_acc_history, val_acc_history


def plot_results(results, title_prefix=""):
    """
    Вспомогательная функция для построения графиков.
    results: словарь {имя_эксперимента: (loss_hist, train_acc, val_acc)}
    """
    fig, axes = plt.subplots(3, 1, figsize=(12, 12))
    for name, (loss_hist, train_acc, val_acc) in results.items():
        axes[0].plot(loss_hist, label=name)
        axes[1].plot(train_acc, label=name)
        axes[2].plot(val_acc, label=name)

    axes[0].set_title(f'{title_prefix} Потери обучения')
    axes[0].set_xlabel('Эпоха')
    axes[0].set_ylabel('Потери')
    axes[0].legend()
    axes[1].set_title(f'{title_prefix} Точность на обучении')
    axes[1].set_xlabel('Эпоха')
    axes[1].set_ylabel('Точность')
    axes[1].legend()
    axes[2].set_title(f'{title_prefix} Точность на валидации')
    axes[2].set_xlabel('Эпоха')
    axes[2].set_ylabel('Точность')
    axes[2].legend()
    plt.tight_layout()
    plt.show()


# ======================================================================
# 6. ЗАПУСК ЭКСПЕРИМЕНТОВ (заполнить по необходимости)
# ======================================================================

def run_experiments():
    # Загрузка данных
    X_train, y_train, X_test, y_test = load_and_prepare_data()

    # Разделение на train и validation
    val_size = 5000
    X_val = X_train[-val_size:]
    y_val = y_train[-val_size:]
    X_train = X_train[:-val_size]
    y_train = y_train[:-val_size]

    # Базовая архитектура
    input_size = 32*32*3
    hidden_sizes = [256, 128]
    output_size = 10
    layer_sizes = [input_size] + hidden_sizes + [output_size]
    activations = ['relu', 'relu', 'softmax']

    # Параметры обучения
    batch_size = 64
    epochs = 30
    learning_rate = 0.001

    # ------------------------------------------------------------------
    # 6.1. Сравнение оптимизаторов (без регуляризации)
    # ------------------------------------------------------------------
    optimizers = ['sgd', 'sgd_momentum', 'rmsprop', 'adam']
    opt_configs = {
        'sgd': {'learning_rate': 0.01},
        'sgd_momentum': {'learning_rate': 0.01, 'momentum': 0.9},
        'rmsprop': {'learning_rate': 0.001},
        'adam': {'learning_rate': 0.001}
    }

    results_opt = {}
    for name in optimizers:
        print(f"\n=== Оптимизатор: {name} ===")
        model = MLP(layer_sizes, activations,
                    optimizer=name, **opt_configs[name])
        loss_hist, train_acc, val_acc = train_model(
            model, X_train, y_train, X_val, y_val,
            batch_size=batch_size, epochs=epochs,
            learning_rate=opt_configs[name]['learning_rate'], verbose=True
        )
        results_opt[name] = (loss_hist, train_acc, val_acc)

    plot_results(results_opt, "Сравнение оптимизаторов")

    # ------------------------------------------------------------------
    # 6.2. Сравнение регуляризаций (используем лучший оптимизатор, например Adam)
    # ------------------------------------------------------------------
    best_optimizer = 'adam'  # замените на лучший по результатам первого эксперимента

    reg_configs = [
        {'name': 'No reg', 'l2': 0.0, 'l1': 0.0, 'dropout': [0.0, 0.0]},
        {'name': 'L2 (0.001)', 'l2': 0.001, 'l1': 0.0, 'dropout': [0.0, 0.0]},
        {'name': 'L1 (0.001)', 'l2': 0.0, 'l1': 0.001, 'dropout': [0.0, 0.0]},
        {'name': 'Dropout (0.3,0.3)', 'l2': 0.0, 'l1': 0.0, 'dropout': [0.3, 0.3]},
        {'name': 'L2+Dropout', 'l2': 0.001, 'l1': 0.0, 'dropout': [0.3, 0.3]},
    ]

    results_reg = {}
    for config in reg_configs:
        print(f"\n--- {config['name']} ---")
        model = MLP(layer_sizes, activations,
                    dropout_probs=config['dropout'] + [0.0],  # выходной слой без dropout
                    l2_lambda=config['l2'],
                    l1_lambda=config['l1'],
                    optimizer=best_optimizer,
                    learning_rate=0.001)
        loss_hist, train_acc, val_acc = train_model(
            model, X_train, y_train, X_val, y_val,
            batch_size=batch_size, epochs=epochs,
            learning_rate=0.001, verbose=True
        )
        results_reg[config['name']] = (loss_hist, train_acc, val_acc)

    plot_results(results_reg, "Сравнение регуляризаций")

    # Дополнительно можно вывести итоговые точности на тесте
    print("\n=== Итоговая точность на тестовой выборке ===")
    for name, (_, _, _) in results_opt.items():
        # Для оптимизаторов нужно заново обучить или использовать последнюю модель, но для простоты не будем.
        pass


if __name__ == "__main__":
    run_experiments()
```

```python
"""
ЛР 1.3. Реализация SOM для визуализации данных (шаблон для студента)

Цель: Изучить архитектуру SOM и применить её для визуализации многомерных данных.

Датасет: Wine (13 химических показателей, 3 класса вин)

Задачи:
1. Реализовать алгоритм обучения SOM с нуля на NumPy.
2. Обучить SOM на датасете Wine.
3. Визуализировать полученную карту признаков с цветовым кодированием кластеров.
4. Интерпретировать результаты: показать, как данные группируются на карте.

Технологии: Python, NumPy, Matplotlib, scikit-learn.
"""

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')


# ================== 1. Реализация SOM (задание для студента) ==================
class SOM:
    """
    Самоорганизующаяся карта Кохонена (SOM).
    
    Параметры:
    - m, n: размеры сетки (количество нейронов по вертикали и горизонтали)
    - dim: размерность входных данных (количество признаков)
    - lr: начальная скорость обучения (learning rate)
    - sigma: начальный радиус соседства (sigma)
    - random_seed: для воспроизводимости результатов
    - metric: метрика расстояния ('euclidean' или 'cosine')
    """
    def __init__(self, m, n, dim, lr=0.5, sigma=1.0, random_seed=42, metric='euclidean'):
        self.m = m
        self.n = n
        self.dim = dim
        self.initial_lr = lr
        self.initial_sigma = sigma
        self.metric = metric
        
        np.random.seed(random_seed)
        self.weights = None
        self.initialized = False
        
    def _initialize_weights(self, X):
        """
        Инициализация весов с помощью случайных выборок из данных.
        Это улучшает сходимость по сравнению со случайной инициализацией.
        """
        n_samples = X.shape[0]
        indices = np.random.choice(n_samples, size=self.m * self.n, replace=True)
        self.weights = X[indices].reshape(self.m, self.n, self.dim)
        self.initialized = True
        
    def _find_bmu(self, x):
        """
        TODO: Найти нейрон-победитель (Best Matching Unit) для вектора x.
        
        Подсказки:
        1. Преобразуйте веса self.weights в плоский массив размером (m*n, dim)
        2. Вычислите расстояние от x до каждого нейрона в зависимости от self.metric:
           - Для 'euclidean': евклидово расстояние (сумма квадратов разностей)
           - Для 'cosine': косинусное расстояние (1 - косинусное сходство)
        3. Найдите индекс нейрона с минимальным расстоянием
        4. Преобразуйте индекс в координаты (i, j) на сетке
        5. Верните (i, j, минимальное_расстояние)
        
        Примечание: используйте np.argmin() для поиска индекса минимума.
        """
        # TODO: Реализуйте поиск нейрона-победителя
        pass
    
    def _get_neighborhood(self, bmu_i, bmu_j, sigma):
        """
        TODO: Вычислить коэффициенты соседства для всех нейронов относительно BMU.
        
        Подсказки:
        1. Создайте матрицы координат i_coords (m, 1) и j_coords (1, n)
        2. Вычислите квадрат расстояния от каждого нейрона до BMU:
           dist_sq = (i_coords - bmu_i)**2 + (j_coords - bmu_j)**2
        3. Примените гауссово окно: exp(-dist_sq / (2 * sigma**2))
        4. Верните матрицу коэффициентов соседства размером (m, n)
        
        Примечание: используйте np.exp() для вычисления экспоненты.
        """
        # TODO: Реализуйте вычисление коэффициентов соседства
        pass
    
    def fit(self, X, epochs, verbose=True, X_test=None):
        """
        TODO: Реализовать обучение SOM.
        
        Алгоритм:
        Для каждой эпохи:
            1. Вычислить текущие lr и sigma (экспоненциальное затухание)
            2. Перемешать данные (np.random.permutation)
            3. Для каждого образца:
               a. Найти BMU (использовать _find_bmu)
               b. Вычислить коэффициенты соседства (использовать _get_neighborhood)
               c. Обновить веса всех нейронов:
                  delta = lr * neighborhood * (x - weights)
                  weights += delta
            4. Вычислить ошибку на эпохе (использовать _compute_quantization_error)
            5. Выводить прогресс каждые 10% эпох
        
        Подсказки:
        - Используйте экспоненциальное затухание: lr = initial_lr * exp(-epoch / epochs)
        - Веса имеют форму (m, n, dim), neighborhood - (m, n)
        - Для broadcasting добавьте новую ось: neighborhood[..., np.newaxis]
        - Для перемешивания: indices = np.random.permutation(n_samples)
        
        Параметры:
        - X: массив формы (n_samples, dim) - входные данные
        - epochs: количество эпох обучения
        - verbose: печатать прогресс
        - X_test: тестовые данные для оценки качества (опционально)
        """
        # TODO: Реализуйте основной цикл обучения SOM
        pass
    
    def _compute_quantization_error(self, X):
        """
        Вычисляет среднюю ошибку квантования (среднее расстояние до BMU).
        """
        # TODO: Вычислите среднюю ошибку квантования
        pass
    
    def predict(self, X):
        """
        Для каждого образца в X возвращает координаты BMU на карте.
        Возвращает массив координат (i, j) формы (n_samples, 2).
        """
        # TODO: Реализуйте предсказание координат BMU
        pass
    
    def get_weights(self):
        """Возвращает веса нейронов."""
        return self.weights
    
    def get_component_planes(self):
        """
        Возвращает компонентные плоскости для каждого признака.
        Каждая плоскость показывает, как меняется вес данного признака по карте.
        """
        planes = []
        for d in range(self.dim):
            plane = self.weights[:, :, d]
            # Нормализуем для лучшей визуализации
            plane = (plane - plane.min()) / (plane.max() - plane.min() + 1e-10)
            planes.append(plane)
        return planes


# ================== 2. Подготовка данных ==================
# Загружаем датасет Wine
print("Загрузка датасета Wine...")
wine = load_wine()
X_wine = wine.data
y_wine = wine.target
feature_names = wine.feature_names
target_names = wine.target_names

print(f"Размерность данных: {X_wine.shape}")
print(f"Количество классов: {len(target_names)}")
print(f"Признаки: {feature_names}")

# Стандартизация данных (критически важно для SOM)
scaler = StandardScaler()
X_wine_scaled = scaler.fit_transform(X_wine)

# Разделяем на обучающую и тестовую выборки
np.random.seed(42)
indices = np.random.permutation(len(X_wine_scaled))
split_idx = int(0.8 * len(X_wine_scaled))
X_train = X_wine_scaled[indices[:split_idx]]
y_train = y_wine[indices[:split_idx]]
X_test = X_wine_scaled[indices[split_idx:]]
y_test = y_wine[indices[split_idx:]]

print(f"Обучающая выборка: {X_train.shape[0]} образцов")
print(f"Тестовая выборка: {X_test.shape[0]} образцов")


# ================== 3. Обучение SOM ==================
# Параметры SOM
SOM_SIZE = (15, 15)  # 15x15 нейронов
DIM = X_wine_scaled.shape[1]  # 13 признаков

# Создаём и обучаем SOM на Wine
som_wine = SOM(
    m=SOM_SIZE[0], 
    n=SOM_SIZE[1], 
    dim=DIM, 
    lr=0.5, 
    sigma=3.0,
    metric='euclidean'
)

print("\nОбучение SOM на датасете Wine...")
som_wine.fit(X_train, epochs=300, verbose=True, X_test=X_test)


# ================== 4. Визуализация (изучите функции, отбражения карт и компонентных плоскостей) ==================
# Определяем цвета для классов (красный, синий, зелёный)
CLASS_COLORS = ['red', 'blue', 'green']

# Проверяем доступные стили
try:
    plt.style.use('seaborn-v0_8-darkgrid')
except OSError:
    try:
        plt.style.use('seaborn-darkgrid')
    except OSError:
        try:
            plt.style.use('ggplot')
        except OSError:
            pass

def plot_training_errors(som, title="Кривая обучения SOM"):
    """Визуализация ошибки обучения."""
    plt.figure(figsize=(10, 5))
    epochs = range(1, len(som.train_errors) + 1)
    plt.plot(epochs, som.train_errors, 'b-', label='Ошибка на обучении', linewidth=2)
    if som.test_errors is not None:
        plt.plot(epochs, som.test_errors, 'r--', label='Ошибка на тесте', linewidth=2)
    plt.xlabel('Эпоха')
    plt.ylabel('Средняя ошибка квантования')
    plt.title(title, fontsize=14)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

def plot_som_with_labels(som, X, y, title="SOM визуализация", class_colors=None):
    """
    Визуализирует карту SOM с цветовым кодированием кластеров.
    """
    if class_colors is None:
        class_colors = ['red', 'blue', 'green']
    
    bmu_coords, distances = som.predict(X)
    
    fig, ax = plt.subplots(figsize=(12, 10))
    
    # Добавляем небольшой шум для лучшей визуализации
    jitter_strength = 0.3
    for i, (coord, label) in enumerate(zip(bmu_coords, y)):
        x_pos = coord[1] + np.random.uniform(-jitter_strength, jitter_strength)
        y_pos = coord[0] + np.random.uniform(-jitter_strength, jitter_strength)
        # Размер точки зависит от ошибки
        size = 20 + 30 * (distances[i] / (max(distances) + 1e-10))
        color = class_colors[int(label) % len(class_colors)]
        ax.scatter(x_pos, y_pos, color=color, s=size, alpha=0.8, edgecolors='white', linewidth=0.5)
    
    # Настраиваем оси
    ax.set_xlim(-0.5, som.n - 0.5)
    ax.set_ylim(-0.5, som.m - 0.5)
    ax.set_xticks(range(0, som.n, 2))
    ax.set_yticks(range(0, som.m, 2))
    ax.grid(True, linestyle='--', alpha=0.3)
    ax.set_title(title, fontsize=16, pad=20)
    ax.set_xlabel("Координата по горизонтали", fontsize=12)
    ax.set_ylabel("Координата по вертикали", fontsize=12)
    
    # Легенда
    unique_labels = np.unique(y)
    legend_handles = []
    for label in unique_labels:
        handle = plt.Line2D([0], [0], marker='o', color='w',
                           markerfacecolor=class_colors[int(label) % len(class_colors)],
                           markersize=12, label=f'{target_names[label]}')
        legend_handles.append(handle)
    ax.legend(handles=legend_handles, loc='upper right', fontsize=11)
    
    plt.tight_layout()
    plt.show()

def plot_component_planes(som, feature_names, title="Компонентные плоскости SOM"):
    """Визуализирует компонентные плоскости для каждого признака."""
    planes = som.get_component_planes()
    n_features = len(planes)
    cols = 4
    rows = (n_features + cols - 1) // cols
    
    fig, axes = plt.subplots(rows, cols, figsize=(16, 4 * rows))
    axes = axes.flatten()
    
    for idx, (plane, name) in enumerate(zip(planes, feature_names)):
        im = axes[idx].imshow(plane, cmap='viridis', origin='upper', aspect='auto')
        axes[idx].set_title(name, fontsize=10)
        axes[idx].axis('off')
        plt.colorbar(im, ax=axes[idx], fraction=0.046, pad=0.04)
    
    for idx in range(n_features, len(axes)):
        axes[idx].axis('off')
    
    plt.suptitle(title, fontsize=16, y=1.02)
    plt.tight_layout()
    plt.show()

def plot_umatrix(som, title="U-Matrix (карта расстояний)"):
    """Визуализирует U-Matrix - среднее расстояние между соседними нейронами."""
    weights = som.get_weights()
    m, n, dim = weights.shape
    
    umatrix = np.zeros((m, n))
    for i in range(m):
        for j in range(n):
            neighbors = []
            if i > 0:
                neighbors.append(weights[i-1, j])
            if i < m-1:
                neighbors.append(weights[i+1, j])
            if j > 0:
                neighbors.append(weights[i, j-1])
            if j < n-1:
                neighbors.append(weights[i, j+1])
            if neighbors:
                dists = [np.linalg.norm(weights[i, j] - neigh) for neigh in neighbors]
                umatrix[i, j] = np.mean(dists)
    
    umatrix = (umatrix - umatrix.min()) / (umatrix.max() - umatrix.min() + 1e-10)
    
    plt.figure(figsize=(10, 8))
    plt.imshow(umatrix, cmap='hot_r', origin='upper', interpolation='bilinear')
    plt.colorbar(label='Среднее расстояние до соседей (нормализовано)')
    plt.title(title, fontsize=16)
    plt.xlabel("Координата по горизонтали", fontsize=12)
    plt.ylabel("Координата по вертикали", fontsize=12)
    plt.tight_layout()
    plt.show()

def plot_hit_map(som, X, y, title="Hit Map (плотность попаданий)", class_colors=None):
    """Визуализирует плотность попаданий данных в нейроны."""
    if class_colors is None:
        class_colors = ['red', 'blue', 'green']
    
    bmu_coords, _ = som.predict(X)
    
    hit_map = np.zeros((som.m, som.n))
    for coord in bmu_coords:
        hit_map[coord[0], coord[1]] += 1
    
    if hit_map.max() > 0:
        hit_map = hit_map / hit_map.max()
    
    plt.figure(figsize=(10, 8))
    plt.imshow(hit_map, cmap='Blues', origin='upper', interpolation='bilinear')
    plt.colorbar(label='Нормализованная плотность попаданий')
    plt.title(title, fontsize=16)
    plt.xlabel("Координата по горизонтали", fontsize=12)
    plt.ylabel("Координата по вертикали", fontsize=12)
    
    for label in np.unique(y):
        mask = y == label
        coords = bmu_coords[mask]
        if len(coords) > 0:
            mean_i = np.mean(coords[:, 0])
            mean_j = np.mean(coords[:, 1])
            color = class_colors[int(label) % len(class_colors)]
            plt.plot(mean_j, mean_i, marker='*', markersize=15, 
                    color=color, label=f'{target_names[label]} (центр)')
    
    plt.legend(loc='upper right')
    plt.tight_layout()
    plt.show()


# ================== 5. Построение визуализаций ==================
print("\n" + "="*60)
print("ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ")
print("="*60)

# 1. Кривая обучения
print("\n1. Кривая обучения SOM...")
plot_training_errors(som_wine, title="Кривая обучения SOM на датасете Wine")

# 2. Карта SOM с цветовым кодированием (обучающая выборка)
print("\n2. Карта SOM для обучающей выборки...")
plot_som_with_labels(
    som_wine, X_train, y_train, 
    title="SOM визуализация: Wine (обучающая выборка)",
    class_colors=CLASS_COLORS
)

# 3. Карта SOM для тестовой выборки
print("\n3. Карта SOM для тестовой выборки...")
plot_som_with_labels(
    som_wine, X_test, y_test, 
    title="SOM визуализация: Wine (тестовая выборка)",
    class_colors=CLASS_COLORS
)

# 4. U-Matrix
print("\n4. U-Matrix (карта расстояний)...")
plot_umatrix(som_wine, title="U-Matrix: Wine")

# 5. Hit Map
print("\n5. Hit Map (плотность попаданий)...")
plot_hit_map(
    som_wine, X_train, y_train, 
    title="Hit Map: Wine (обучение)",
    class_colors=CLASS_COLORS
)

# 6. Компонентные плоскости
print("\n6. Компонентные плоскости для всех 13 признаков...")
plot_component_planes(som_wine, feature_names, title="Компонентные плоскости SOM для Wine")


# ================== 6. Интерпретация результатов ==================
print("\n" + "="*60)
print("ИНТЕРПРЕТАЦИЯ РЕЗУЛЬТАТОВ")
print("="*60)

print(f"""
1. КАРТА SOM (цвета: {', '.join(CLASS_COLORS)}):
   - На карте чётко видны три кластера, соответствующие трём сортам вина.
   - Кластер 0 ({target_names[0]}) отображается КРАСНЫМ цветом.
   - Кластер 1 ({target_names[1]}) отображается СИНИМ цветом.
   - Кластер 2 ({target_names[2]}) отображается ЗЕЛЁНЫМ цветом.
   - Кластер 0 обычно изолирован от двух других.
   - Кластеры 1 и 2 могут частично перекрываться, но SOM разделяет их.

2. U-MATRIX:
   - Тёмные области (низкие значения) соответствуют кластерам.
   - Светлые линии (высокие значения) показывают границы между кластерами.
   - Помогает визуально определить количество и структуру кластеров.

3. HIT MAP:
   - Показывает, в какие нейроны чаще всего попадают данные.
   - Центры кластеров отмечены звёздочками соответствующих цветов.
   - Можно оценить плотность распределения данных на карте.

4. КОМПОНЕНТНЫЕ ПЛАНЫ:
   - Для каждого химического признака показывают, как его значение 
     меняется по карте.
   - Признаки с похожими паттернами могут быть коррелированы.
   - Помогают интерпретировать, какие признаки важны для разделения классов.

5. КРИВАЯ ОБУЧЕНИЯ:
   - Показывает сходимость алгоритма.
   - Ошибка на тесте должна быть близка к ошибке на обучении, 
     что говорит о хорошей обобщающей способности.

ВЫВОДЫ:
- SOM успешно визуализирует 13-мерное пространство на 2D карте.
- Топологическое свойство SOM сохраняется: близкие образцы 
  отображаются в соседние нейроны.
- Компонентные плоскости позволяют интерпретировать, какие 
  химические признаки являются наиболее дискриминантными.
- Метод может быть использован для исследования структуры 
  многомерных данных и выявления скрытых закономерностей.
""")

```

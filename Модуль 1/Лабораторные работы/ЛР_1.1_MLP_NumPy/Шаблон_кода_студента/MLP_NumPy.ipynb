```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical

# ================== ЗАДАНИЕ 1: РЕАЛИЗУЙТЕ КЛАСС LAYER ==================
class Layer:
    """
    Один полносвязный слой с активацией.
    """

    def __init__(self, n_input, n_output, activation='relu'):
        """
        Инициализация весов и смещений.
        :param n_input:  размер входного вектора
        :param n_output: количество нейронов
        :param activation: 'relu', 'softmax' или 'linear'
        """
        # TODO: Инициализируйте W и b.
        #   - Для relu используйте He-инициализацию: W ~ N(0, sqrt(2/n_input))
        #   - Для других активаций – малые случайные числа (например, 0.01 * randn)
        #   - b инициализируйте нулями (размер 1 x n_output)
        #   - Сохраните activation
        pass

        # Поля для хранения промежуточных данных
        self.input = None   # вход X
        self.Z = None       # линейная часть X·W + b
        self.output = None  # результат активации

    def forward(self, X):
        """
        Прямой проход: Z = X·W + b, затем активация.
        :param X: входной батч (batch_size, n_input)
        :return: выход после активации
        """
        # TODO: 
        # 1. Сохраните X в self.input
        # 2. Вычислите Z = X·W + b, сохраните в self.Z
        # 3. В зависимости от self.activation:
        #    - 'relu': output = max(0, Z)
        #    - 'softmax': стабильный softmax (вычтите максимум из Z по строкам)
        #    - 'linear': output = Z
        # 4. Сохраните output в self.output и верните его
        pass

    def backward(self, dout, learning_rate):
        """
        Обратный проход: вычисление градиентов и обновление параметров.
        :param dout: градиент по выходу слоя (dLoss / dOutput)
        :param learning_rate: скорость обучения
        :return: градиент по входу (dLoss / dInput) для предыдущего слоя
        """
        # TODO:
        # 1. Вычислите dZ = dout * производная_активации(Z)
        #    Для ReLU: dZ = dout * (Z > 0)
        #    Для softmax: dZ = dout (т.к. градиент уже содержит (prob - y_true)/batch_size)
        #    Для linear: dZ = dout
        #
        # 2. Вычислите градиенты по параметрам (усредните по батчу):
        #    dW = (X.T · dZ) / batch_size
        #    db = sum(dZ, axis=0, keepdims=True) / batch_size
        #
        # 3. Вычислите градиент по входу для предыдущего слоя:
        #    dX = dZ · W.T
        #
        # 4. Обновите веса: W -= learning_rate * dW, b -= learning_rate * db
        #
        # 5. Верните dX
        pass


# ================== ЗАДАНИЕ 2: РЕАЛИЗУЙТЕ КЛАСС MLP ==================
class MLP:
    """
    Многослойный перцептрон, состоящий из слоёв Layer.
    """

    def __init__(self, layer_sizes, activations):
        """
        :param layer_sizes: список размеров слоёв, например [3072, 256, 128, 10]
        :param activations: список активаций для каждого слоя (длина = len(layer_sizes)-1)
        """
        # TODO: Создайте список self.layers, в котором будут экземпляры Layer.
        #   Для каждой пары (n_input, n_output) из layer_sizes создайте слой.
        pass

    def forward(self, X):
        """Прямой проход через все слои."""
        # TODO: Прогоните X через все слои и верните выход последнего.
        pass

    def backward(self, dout, learning_rate):
        """Обратный проход через все слои (от последнего к первому)."""
        # TODO: Передайте градиент dout через слои в обратном порядке,
        #   вызывая для каждого layer.backward().
        pass

    def train_step(self, X_batch, y_batch, learning_rate):
        """
        Один шаг обучения на батче.
        :param X_batch: входные данные (batch_size, input_dim)
        :param y_batch: метки классов (индексы, 0..9)
        :param learning_rate: скорость обучения
        :return: значение кросс-энтропии на этом батче
        """
        # TODO:
        # 1. Прямой проход: получите вероятности probs.
        # 2. Преобразуйте y_batch в one-hot (используйте to_categorical).
        # 3. Вычислите потери: -mean( sum( y_one_hot * log(probs + 1e-8) ) )
        # 4. Вычислите градиент для выходного слоя: dout = (probs - y_one_hot) / batch_size
        # 5. Вызовите self.backward(dout, learning_rate)
        # 6. Верните значение потерь
        pass

    def predict(self, X):
        """Предсказание классов (индексы с максимальной вероятностью)."""
        # TODO: Верните argmax по выходу forward(X)
        pass

    def accuracy(self, X, y):
        """Точность на выборке."""
        # TODO: Сравните предсказания с истинными метками, верните среднюю долю правильных.
        pass


# ================== ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ (НЕ МЕНЯТЬ) ==================

def visualize_samples(X_raw, y_raw, num_examples=5, figsize=(15, 8)):
    """Визуализация примеров (по одному из каждого класса в строке)."""
    class_names_ru = ['Самолёт', 'Автомобиль', 'Птица', 'Кошка', 'Олень',
                      'Собака', 'Лягушка', 'Лошадь', 'Корабль', 'Грузовик']
    fig, axes = plt.subplots(num_examples, 10, figsize=figsize,
                             subplot_kw={'xticks': [], 'yticks': []})
    fig.subplots_adjust(hspace=0.1, wspace=0.1)
    for row in range(num_examples):
        for class_id in range(10):
            idxs = np.where(y_raw == class_id)[0]
            idx = np.random.choice(idxs)
            img = X_raw[idx].astype(np.float32) / 255.0
            ax = axes[row, class_id]
            ax.imshow(img, interpolation='nearest')
            ax.axis('off')
            if row == 0:
                ax.set_title(class_names_ru[class_id], fontsize=8)
    plt.suptitle(f'Примеры изображений CIFAR-10 (по {num_examples} случайных на класс)',
                 fontsize=14)
    plt.show()

def load_and_prepare_data():
    """Загрузка и нормализация данных."""
    (X_train, y_train), (X_test, y_test) = cifar10.load_data()
    X_train_raw = X_train.copy()
    X_test_raw = X_test.copy()
    X_train = X_train.reshape(X_train.shape[0], -1) / 255.0
    X_test  = X_test.reshape(X_test.shape[0], -1) / 255.0
    y_train = y_train.flatten()
    y_test  = y_test.flatten()
    # Стандартизация (опционально, можете добавить)
    # mean = X_train.mean(axis=0)
    # std = X_train.std(axis=0) + 1e-8
    # X_train = (X_train - mean) / std
    # X_test = (X_test - mean) / std
    return X_train, y_train, X_test, y_test, X_train_raw, X_test_raw


# ================== ОБУЧЕНИЕ (НЕ МЕНЯТЬ) ==================

def train_mlp():
    # Загружаем данные
    X_train, y_train, X_test, y_test, X_train_raw, X_test_raw = load_and_prepare_data()

    # Визуализация примеров
    visualize_samples(X_train_raw, y_train, num_examples=5)

    # Параметры модели (вы можете их изменять для улучшения точности)
    input_size = 32 * 32 * 3
    hidden_sizes = [256, 128]           # попробуйте увеличить
    output_size = 10
    layer_sizes = [input_size] + hidden_sizes + [output_size]
    activations = ['relu', 'relu', 'softmax']  # для скрытых – ReLU, для выхода – softmax

    learning_rate = 0.001
    batch_size = 64
    epochs = 30

    # Создание модели
    model = MLP(layer_sizes, activations)

    train_loss_history = []
    train_acc_history = []
    test_acc_history = []

    num_batches = int(np.ceil(X_train.shape[0] / batch_size))

    print("Начинаем обучение...")
    for epoch in range(epochs):
        # Перемешивание
        indices = np.random.permutation(X_train.shape[0])
        X_train_shuffled = X_train[indices]
        y_train_shuffled = y_train[indices]

        epoch_loss = 0.0
        for batch_idx in range(num_batches):
            start = batch_idx * batch_size
            end = min(start + batch_size, X_train.shape[0])
            X_batch = X_train_shuffled[start:end]
            y_batch = y_train_shuffled[start:end]

            loss = model.train_step(X_batch, y_batch, learning_rate)
            epoch_loss += loss * (end - start)

        epoch_loss /= X_train.shape[0]
        train_loss_history.append(epoch_loss)

        train_acc = model.accuracy(X_train, y_train)
        train_acc_history.append(train_acc)
        test_acc = model.accuracy(X_test, y_test)
        test_acc_history.append(test_acc)

        print(f"Эпоха {epoch+1}/{epochs}  Loss: {epoch_loss:.4f}  "
              f"Train Acc: {train_acc:.4f}  Test Acc: {test_acc:.4f}")

    # Визуализация графиков
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(train_loss_history, label='Train Loss')
    plt.xlabel('Эпоха')
    plt.ylabel('Потери (кросс-энтропия)')
    plt.title('График потерь')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(train_acc_history, label='Train Accuracy')
    plt.plot(test_acc_history, label='Test Accuracy')
    plt.xlabel('Эпоха')
    plt.ylabel('Точность')
    plt.title('График точности')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    train_mlp()
```
